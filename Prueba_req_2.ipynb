```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual simple
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11

# Cargar datos
df = pd.read_csv("05. transformacion_digital_dataset.csv")

# Revisar estructura
print(df.head())
print(df.info())
print(df.describe())
```

```python
print(df["mes"].head(10))
df["mes"] = pd.to_datetime(df["mes"])
df = df.sort_values("mes")
```

```python
plt.figure()
plt.plot(df["mes"], df["productividad_manufactura"], marker="o", label="Productividad")
plt.plot(df["mes"], df["nps_score"], marker="o", label="NPS")
plt.plot(df["mes"], df["recompra_porcentaje"], marker="o", label="Recompra")
plt.title("La transformación mejora productividad, NPS y recompra")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure()
plt.plot(df["mes"], df["tasa_defectos_porcentaje"], marker="o", label="Defectos %")
plt.plot(df["mes"], df["tiempo_ciclo_horas"], marker="o", label="Tiempo ciclo")
plt.plot(df["mes"], df["tasa_reclamos_porcentaje"], marker="o", label="Reclamos %")
plt.title("Los principales indicadores operativos muestran mejora")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure()
plt.plot(df["mes"], df["empleados_capacitados_porcentaje"], marker="o", label="Capacitados %")
plt.plot(df["mes"], df["procesos_digitalizados_porcentaje"], marker="o", label="Procesos digitalizados %")
plt.plot(df["mes"], df["uso_plataforma_digital_porcentaje"], marker="o", label="Uso plataforma %")
plt.plot(df["mes"], df["roi_transformacion_porcentaje"], marker="o", label="ROI %")
plt.title("La adopción digital avanza y el ROI mejora")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure()
plt.plot(df["mes"], df["horas_extra_promedio"], marker="o", label="Horas extra")
plt.plot(df["mes"], df["rotacion_personal_porcentaje"], marker="o", label="Rotación %")
plt.plot(df["mes"], df["vacantes_tech_sin_cubrir"], marker="o", label="Vacantes tech")
plt.plot(df["mes"], df["satisfaccion_empleados_score"], marker="o", label="Satisfacción empleados")
plt.title("La transformación también genera presión sobre las personas")
plt.xlabel("Mes")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
